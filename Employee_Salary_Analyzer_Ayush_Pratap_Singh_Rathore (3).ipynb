## what does this project do ?
To organize employee salary information and perform simple analysis:
- Calculate total salary (basic salary + bonus)
- Find average, highest and lowest salary
- Compare average salaries across departments
- Display simple charts





## 1. Import the libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

print("Libraries imported successfully!")
```

## 2. Create a sample employee dataset

```python
data = {
    "Employee_ID": [101, 102, 103, 104, 105, 106, 107, 108],
    "Employee_Name": ["Aarav", "Priya", "Rahul", "Sneha", "Karan", "Ananya", "Rohit", "Meera"],
    "Department": ["IT", "HR", "IT", "Finance", "Sales", "HR", "Finance", "Sales"],
    "Basic_Salary": [30000, 28000, 35000, 32000, 25000, 31000, 36000, 27000],
    "Bonus": [3000, 2000, 4000, 2500, 1500, 2500, 3500, 2000]
}

df = pd.DataFrame(data)
df.to_csv("employee_salary_data.csv", index=False)

print("Sample employee dataset created!")
display(df)
```

## 3. Read the CSV file using Pandas

```python
df = pd.read_csv("employee_salary_data.csv")
print("Dataset loaded successfully!")
display(df.head())
```

## 4. Calculate total salary

```python
# Total Salary = Basic Salary + Bonus
df["Total_Salary"] = df["Basic_Salary"] + df["Bonus"]

display(df[["Employee_Name", "Department", "Basic_Salary", "Bonus", "Total_Salary"]])
```

## 5. Salary statistics using Pandas and NumPy

```python
print("Total salary amount:", df["Total_Salary"].sum())
print("Average salary:", round(df["Total_Salary"].mean(), 2))
print("Highest salary:", df["Total_Salary"].max())
print("Lowest salary:", df["Total_Salary"].min())

# Convert salary column into a NumPy array
salary_values = df["Total_Salary"].to_numpy()
print("NumPy average:", round(np.mean(salary_values), 2))
print("NumPy standard deviation:", round(np.std(salary_values), 2))
```

## 6. Find highest-paid and lowest-paid employees

```python
highest_paid = df.loc[df["Total_Salary"].idxmax()]
lowest_paid = df.loc[df["Total_Salary"].idxmin()]

print("Highest-paid employee:")
display(highest_paid.to_frame().T)

print("Lowest-paid employee:")
display(lowest_paid.to_frame().T)
```

## 7. Department-wise average salary

```python
department_salary = df.groupby("Department")["Total_Salary"].mean().round(2)
display(department_salary)
```

## 8. Filter employees earning more than 35,000

```python
high_salary = df[df["Total_Salary"] > 35000]
display(high_salary[["Employee_Name", "Department", "Total_Salary"]])
```

## 9. Create charts

```python
# Chart 1: Total salary of each employee
plt.figure(figsize=(9, 5))
plt.bar(df["Employee_Name"], df["Total_Salary"])
plt.title("Total Salary of Employees")
plt.xlabel("Employee Name")
plt.ylabel("Total Salary")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Chart 2: Average salary in each department
plt.figure(figsize=(8, 5))
department_salary.plot(kind="bar")
plt.title("Average Salary by Department")
plt.xlabel("Department")
plt.ylabel("Average Total Salary")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

## 10. Project summary

```python
print("EMPLOYEE SALARY ANALYZER - SUMMARY")
print("-----------------------------------")
print("Number of employees:", len(df))
print("Number of departments:", df["Department"].nunique())
print("Total salary amount:", df["Total_Salary"].sum())
print("Average salary:", round(df["Total_Salary"].mean(), 2))
print("Highest salary:", df["Total_Salary"].max())
print("Lowest salary:", df["Total_Salary"].min())
print("Department with highest average salary:", department_salary.idxmax())
```
